```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
orders=pd.read_csv("../data/raw/olist_orders_dataset.csv")
orders_items=pd.read_csv("../data/raw/olist_order_items_dataset.csv")
customers=pd.read_csv("../data/raw/olist_customers_dataset.csv")
products=pd.read_csv("../data/raw/olist_products_dataset.csv")
order_payments=pd.read_csv("../data/raw/olist_order_payments_dataset.csv")
order_reviews=pd.read_csv("../data/raw/olist_order_reviews_dataset.csv")
```

```python
datasets={
    "Orders":orders,
    "Order Items":orders_items,
    "Customers":customers,
    "Products":products,
    "Order payments":order_payments,
    "Order reviews":order_reviews
    }
for names,df in datasets.items():
    print(f"\n{names}")
    print(df.isnull().sum())
```

```python
for names,df in datasets.items():
    print(f"\n{names}")
    missing=df.isnull().sum()
    missing_percentage=(df.isnull().mean()*100)

    missing_analysis=pd.DataFrame({
        "Missing_count":missing,
        "Missing_percentage":missing_percentage
    })
    print(missing_analysis[missing_analysis["Missing_count"]>0])
```

```python
for names,df in datasets.items():
    print(f"\n{names}")
    print(df.duplicated().sum())
```

```python
print("Orders duplicates IDs:",orders["order_id"].duplicated().sum())
print("Order Items duplicates:",orders_items.duplicated(subset=["order_id","order_item_id"]).sum())
print("customer duplicates IDs:",customers["customer_id"].duplicated().sum())
print("Product IDs:",products["product_id"].duplicated().sum())
print("Order Payments:",order_payments["order_id"].duplicated().sum())
```
